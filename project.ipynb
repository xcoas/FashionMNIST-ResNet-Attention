```python
from torchvision import datasets, transforms
from NetworkArchitecture import Neural_FashionMNIST
import torch.nn as nn
import torch
```

```python
train_transform = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor(),
])
```

```python
training_data = datasets.FashionMNIST(
    root="data",
    train=True,
    download=True,
    transform=train_transform,
)

test_data = datasets.FashionMNIST(
    root="data",
    train=False,
    download=True,
    transform=transforms.ToTensor(),
)
```

```python
model = Neural_FashionMNIST()
```

```python
data_loader = torch.utils.data.DataLoader(
    training_data,
    512,
    shuffle=True,
)
```

```python
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.00005, weight_decay=1e-4)
```

```python
test_loader = torch.utils.data.DataLoader(
    test_data,
    batch_size=512,
    shuffle=True
)
```

```python
EPOCHS = 100

best_loss = float('inf')
no_best_repeat = 0

load_model = False

if load_model:
    model.load_state_dict(torch.load("model.pt", weights_only=True))

for epoch in range(EPOCHS):
    for batch, (X, y) in enumerate(data_loader):
        pred = model(X)[0]
        loss = loss_fn(pred, y)

        optimizer.zero_grad()

        loss.backward()
        optimizer.step()

    with torch.no_grad():
        test_loss = 0
        for batch, (X, y) in enumerate(test_loader):
            pred = model(X)[0]
            test_loss += loss_fn(pred, y).item()

        test_loss /= len(test_loader)

        if test_loss < best_loss:
            best_loss = test_loss
            torch.save(model.state_dict(), "model.pt")
            print(f"EPOCH: {epoch} | loss: {test_loss} saved new best model")
            no_best_repeat = 0
        else:
            print(f"EPOCH: {epoch} | loss: {test_loss}")
            if no_best_repeat >= 6:
                break
            no_best_repeat += 1
```
